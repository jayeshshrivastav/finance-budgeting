```python
#Installation of pythin libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import joblib
```

```python
data = pd.read_csv('expense_data_1.csv')
```

```python
data.drop('Subcategory',axis = 1, inplace = True)
```

```python
data.drop('Note.1', axis = 1, inplace = True)
```

```python
data.drop('Account', axis = 1, inplace = True)
```

```python
data.drop('Note', axis = 1, inplace = True)
```

```python
data.drop('INR', axis = 1, inplace = True)
```

```python
data.drop("Currency", axis = 1 , inplace = True)
```

```python
data.drop("Account.1", axis = 1, inplace = True)
```

```python
data.rename(columns = {'Category':'Description','Income/Expense':'Category'},inplace = True)
```

```python
data.head()
```

```python
data["SignedAmount"] = np.where(
    data["Category"] == "Expense",
    -data["Amount"],
    data["Amount"]
)

data["Balance"] = data["SignedAmount"].cumsum()
```

```python
data.head()
```

```python
data["Date"] = pd.to_datetime(data["Date"])

# Aggregate to daily expense
daily_expense = data.groupby(data["Date"].dt.date).agg({
    "Amount": "sum"
}).reset_index()

daily_expense.rename(columns={"Date": "Date", "Amount": "DailyExpense"}, inplace=True)

# ------------------------------
# 2. Feature engineering
# ------------------------------
daily_expense["Date"] = pd.to_datetime(daily_expense["Date"])
daily_expense["DayOfMonth"] = daily_expense["Date"].dt.day
daily_expense["DayOfWeek"] = daily_expense["Date"].dt.weekday
daily_expense["Month"] = daily_expense["Date"].dt.month
daily_expense["Year"] = daily_expense["Date"].dt.year

# Lag features (previous day, previous week)
daily_expense["PrevDayExpense"] = daily_expense["DailyExpense"].shift(1)
daily_expense["PrevWeekExpense"] = daily_expense["DailyExpense"].shift(7)

# Drop NaN (from shifting)
daily_expense = daily_expense.dropna()
```

```python
X = daily_expense[["DayOfMonth", "DayOfWeek", "Month", "PrevDayExpense", "PrevWeekExpense"]]
y = daily_expense["DailyExpense"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
```

```python
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))


new_day = pd.DataFrame([{
    "DayOfMonth": 15,
    "DayOfWeek": 1,        # 0=Monday, 1=Tuesday, ...
    "Month": 3,
    "PrevDayExpense": 300, # put yesterday's expense here
    "PrevWeekExpense": 250 # put same weekday last week's expense
}])

prediction = model.predict(new_day)
print("Predicted Expense for given day:", prediction[0])
```

```python
# Saving the trained model
joblib.dump(model, "expense_predictor.pkl")
print("✅ Model saved as expense_predictor.pkl")
```
